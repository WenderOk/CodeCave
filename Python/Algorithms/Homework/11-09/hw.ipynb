### Джобс 25.1
Нам нужен самый большой делитель не равный 1 и самому числу. который при этом не превышает 4000.<br>
Я думаю что легче пойти с конца (начаная с самого числа) и просто найти делитель который меньше 4000.<br>
Хотя если число например натуральное придется проходить его все, что будет долго.<br>
По эффективности требований не было так что и такое решение подойдет<br>
(Это все первичные размышления. Потом я все по 10 раз еще переделывал и менял мнение)

```python
res = []
for n in range(12345, 12365+1):
    d = 2
    minDiv = n
    while d * d <= n:
        if n % d == 0:
            minDiv = min(minDiv, d)
        d+=1
    if n // minDiv < 4000 and minDiv != n:
        res.append((n, n // minDiv))
print(res)

# В правильности не уверен т.к. у того же Артемия меньше чисел, но по условию задачи все подходит так что пойдет.
# Изначально было вообще тупое условие while d > 1 и при этом с этим условием в ответ попадала пара 12346 и 2,
# а стоило заменить на d * d >= n как оно больше не попадает в ответ. (ниже разобрался что это из-за парных делителей)

# Капец, почти 1 час на 1 задачу...

# Я после того как все сделал и проверял все ли правльно понял что тут вообще не то
# Пришлось переделать. Потом еще дошло что максимальный делаить числа можно получить с помощью минимального делителя
# В общем, теперь точно все правильно.
```

```python
# тут проверял делители
n = int(input())
d = n-1
while d * d >= n:
    if n % d == 0:
        print(d, n // d)
    d-=1
```

Так ну тут по сути почти тоже что и в прошлой задачи за исключенем условий выбора делителей так что можно перееиспользовать алгоритм.<br>
Только что тут ограничение на количество делителей так что придется их как-то считать.

<b>Сложность получилась О(sqrt n)</b><br>

Долго тупил не в состоянии понять почему когда я изменил условие на d * d <= n у меня перестали выводится какие-либо числа.<br> Потом кое-как вместе с ИИшницей разжевал этот метод с парными делителями и понял что мы находим по 2 делителя за итерацию по сути... капец глупым себя чувствую

```python
for n in range(12345, 12354+1):
    d = 2
    s, c = 0, 0
    while d * d <= n:
        if n % d == 0:
            c+=2
            s += d + n // d   
            if c > 2:
                break    
        d+=1
    if c == 2:
        print((n, s), end="  ")

# Изяшного решения без "c" не придумал к сожелению... 
        
```

### Джобс 25.4
Как понимаю под сомножетелями подразумеваются те же парные множители.<br> И их типо нужно вычитать, чтобы разница не была больше 500?<br>
Ну я думаю здесь тоже будет удобнее пойти с конца т.к. у нас максимум то есть самые большие делители. (По итогу все равно пошел с начала)


```python
for n in range(12345, 12457+1):
    d = 2
    dif = 0
    while d * d <= n:
        if n % d == 0:
            dif = max(dif, n // d - d)
        if dif > 0 and dif <= 500:
            print((n, dif), end="  ")
            break
        d+=1

# Как-то многовато чисел в ответе изначально вышло. Хотя вроде все по условию (все составные и разница меньше 500).    
# Потом ИИшница подсказала что я теряю правльные ответы в месте записи dif, так как изначально там не было max. И с ним все заработало
# Единственное что нулевые разницы пришлось фильтровать.

# Уже не таким глупым себя чувствую. Правда чтобы врубиться понадобилось около 3 часов...

        
```

### Джобс 25.2
Так, нам нужно найти делители и пробовать делить их на 3, чтобы посчитать количество четных и нечетных трем.<br> 
А потом еще количество поделить чтобы узнать отношение. Ну за основу здесь все равно будет тот же алгоритм с парными делителями<br> только что дополнительные условия и счетчики добавяться.

<b>Сложность получается О(sqrt n)</b> т.к. мы используем все тот же алгоритм с парными делителями<br>

```python
for n in range(100500, 100700+1):
    d = 1
    c1, c2 = 0, 0
    while d * d <= n:
        if n % d == 0:
            if d % 3 == 0:
                c1+=1
            else:
                c2+=1
            if n // d != d: 
                if n // d % 3 == 0:
                    c1 += 1
                else:
                    c2 += 1
        d+=1
    if c1 > c2 * 2:
        print((n, c1-c2), end="  ")

# Да, много if, но я просто не придумал как еще можно проверять 2 делителя сразу

# Как-то подозрительно много ответов получается.
# Не ну если вычислять разницу целочисленным делением, то там конечно меньше,
# но тогда там возможно деление на ноль, поэтому я и проверяю таким условием c1 > c2*2
# И еще не указано в какую сторону отличается. 
# Так что может отличается как в большую так и в меньшу стороны.

# Потом я перечитал 10 раз условие и понял что там нужно выводить разницу а не тоношение...
# А в случае с тем в какую сторону должно отличается там выходит что нужна только большая сторона
# т.к. в меньшу сторону оно не когда не попадает.

# Похоже пора переучиваться читать...
```

### Делители Евич №12
Ууу ну и намудрили. В целом, вроде ничего сверхестественного если я конечно правильно прочитал задание. Короче надо отобрать числа у которых ровно 3 разных четных делителя и вывести в виде таблицы.

Как и в прошлых алгоритмах <b>Сложность выходит О(sqrt n)</b>

```python
for n in range(123040, 123080+1, 2):
    d = 2
    divs = [] # изначально думал обойтись без списка
    while d * d <= n:
        if n % d == 0:     # Нужно БОЛЬШЕ if'ов
            if d % 2 == 0:
                divs.append(d)
            if n // d != d and n // d % 2 == 0: 
                divs.append(n // d)
        d+=1
    if len(divs) == 3:
        res = sorted(divs) # фу, сортед
        print(n, *res)

# На удивление было не так уж сложно. 
# Возможно это из-за того что условие четкое и понятное


```
