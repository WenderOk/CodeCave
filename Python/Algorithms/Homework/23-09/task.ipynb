## Флавий и бинарный поиск
---

### Удаление чисел

```python
n = int(input())
k = int(input())

iters = 1
pos = n
nums = n

while nums >= k:
    if pos % k == 0:
        print(iters)
        break
    nums -= nums//k
    pos -= pos//k
    iters+=1


```

### Приближенный двоичный поиск

```python
n, k = map(int, input().split())

A = list(map(int, input().split()))
B = list(map(int, input().split()))

def binSearch(x):
    lb = 0
    rb = n-1

    while lb <= rb:
        mid = lb + (rb - lb) // 2

        if x == A[mid] :
            return A[mid]
        elif x > A[mid]:
            lb = mid + 1
        else:
            rb = mid - 1
    
    if x-A[lb] <= A[lb-1]-x:
        return A[lb-1]
    else:
        return A[lb]

for x in B:
    print(binSearch(x))
```

### Бинарный поиск медианного вхождения
Как будто можно было и без 2 функций, которые отличаются только 1 условием...

```python
N = int(input()) 
A = [int(a) for a in input().split()]
M = int(input()) 
B = [int(b) for b in input().split()]
# не понятно на кой мы вообще вводим N и M если они нигде не нужны

def upper_bound(A, x):
    lb = 0
    rb = len(A)

    while lb < rb:
        mid = (lb + rb) // 2
        if A[mid] <= x:
            lb = mid+1
        else:
            rb = mid
    return lb

def lower_bound(A, x):
    lb = 0
    rb = len(A)

    while lb < rb:
        mid = (lb + rb) // 2
        if A[mid] < x:
            lb = mid+1
        else:
            rb = mid
    return lb

res = [0] * len(B)

for i in range(len(B)):
    lb = lower_bound(A, B[i])
    rb = upper_bound(A, B[i])
    if lb != rb:
        res[i] = (lb+1 + rb) // 2
print(*res)

```
