## Алгоритм Евклида
---

### НОД двух чисел
ну тут можно просто использовать то что мы написали на уроке так что не сложно

```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

x, y = map(int, input().split())
print(f"НОД({x}, {y}) = {gcd(x,y)}")
```

### Единичный НОД
Тут казалось бы все тоже не очень сложно но числа из едениц... Удивительным образом обычный алгоритма работает и на них

```python
n, m = map(int, input().split())

res = gcd(n, m)

print("1"*res)
```

### НОД нескольких чисел
Ну со списокм самое очивидное решение. Лучше мне кажется не сделаешь

```python
n = int(input())
nums = list(map(int, input().split()))
res = nums[0]

for n in nums:
    res = gcd(res, n)

print(res)
```

### Триада Евклида
Также со списком

```python
def MultiGcd(nums):
    res = nums[0]
    for i in range(1, len(nums)):
        res = gcd(res, nums[i])
    return res

numbers = list(map(int, input().split()))

print(MultiGcd(numbers))
```

### Прямоугольная игра
Я сначала все никак не мог понять как получаются ответы из примеров, а потом перечитал и дошло что он выкидывает не те которые выбрал а наоборот все остальные и тогда все стало понятно. Опять же надо внимательнее читать

```python
n = int(input())

score = n

while n > 1:
    d = 2
    while n % d != 0:
        d += 1
    n //= d
    score += n

print(score)
```
