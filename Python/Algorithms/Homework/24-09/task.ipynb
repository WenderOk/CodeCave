## Бинарный поиск по ответу
---

### Провода

```python
n, k = map(int, input().split())
wires = []
for i in range(n):
    wires.append(int(input()))

lb = 0
rb = max(wires)

while lb < rb:
    mid = (lb + rb + 1) // 2
    c = 0

    for w in wires:
        c += w // mid

    if c >= k:
        lb = mid
    else:
        rb = mid - 1

print(lb)
```

### Очень легкая задача

```python
n, x, y = map(int, input().split())
first = min(x, y)

lb = 0
rb = n * max(x, y)

while lb < rb:
    mid = lb + (rb - lb) // 2
    c = mid // x + mid // y

    if c >= n - 1:
        rb = mid
    else:
        lb = mid + 1

print(first + lb)
```

### Поиск стороны куба

```python
n = int(input())

lb = 0
rb = n**3+1   # по идее можно было бы взять больше, но и этого вполне должно хватить

while lb + 1 < rb:
    mid = lb + (rb - lb) // 2
    if mid ** 3 <= n: # куб со стороной mid вмещается в куб со стороной n, поэтому сдвигаем левую границу
        lb = mid
    else:
        rb = mid

print(lb)
```
