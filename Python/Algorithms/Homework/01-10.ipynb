## Метод двух указателей

### Натуральные числа вида 2𝑛 × 3𝑚 × 5𝑘
---

```python
numbers = []

# Везде шаг 2 чтобы исключить четные числа
for n in range(1, 30, 2):
    for m in range(1, 20, 2):
        for k in range(1, 15, 2):
            num = 2**n * 3**m * 5**k

            if 400000000 <= num <= 500000000:
                numbers.append(num)
numbers.sort()
print(*numbers, sep="\n")
```

### Книги
---
Валера либо очень быстрый либо у него книги какие-то тонкие, потому что, что это за книга которую можно прочитать за минуту???

```python
n, t = map(int, input().split())
a = [int(x) for x in input().split()]

lp = s = res = 0

for rp in range(n):
    s += a[rp]
    while s > t:
        s -= a[lp]
        lp += 1
    res = max(res, rp - lp + 1)

print(res)
```

### Город Че
---
Интересное применение алгоритмов для решения романтический вопросов...

```python
n, r = map(int, input().split())
dist = [int(x) for x in input().split()]

lp = res = 0

for rp in range(n):
    while lp < rp and dist[rp] - dist[lp] > r:
        lp += 1
    res += lp

print(res)
```

### Два дерева на заданном расстоянии
---

```python
n, D = map(int, input().split())
dist = [int(x) for x in input().split()]
res = [0]
lp = 0

for d in dist:
    res.append(res[-1] + d)

for rp in range(n):
    while res[rp] - res[lp] > D:
        lp += 1

    if res[rp] - res[lp] == D:
        print(lp+1, rp+1)
        break
```
