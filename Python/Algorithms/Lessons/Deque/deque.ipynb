```python
from collections import deque

N, D = map(int, input().split())
k = 0
deq = deque([int(x) for x in input().split()])

for i in range(N):
    if len(deq) >= 2 and deq[0] + deq[-1] <= D:
        deq.popleft()
    if len(deq) > 0:
        deq.pop()
        k+=1

print(k)

```

```python
from collections import deque
N, m = map(int, input().split())
R = [0]*(N-m+1)
deq = deque([])
A = [int(x) for x in input().split()]

for i in range(len(deq)):
    while len(deq) > 0 and deq[-1][0] >= A[i]:
        deq.pop()
    deq.append((A[i], i+m-1))
    if i >= m-1:
        R.append(deq[0][0])
    if i == deq[0][1]:
        deque.popleft()
print(*R)
```
