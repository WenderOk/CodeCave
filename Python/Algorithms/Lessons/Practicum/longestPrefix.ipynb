```python
def prefix(a, b):
    pref = []
    for i in range(len(min(a,b, key=len))):
        if a[i] == b[i]:
            pref.append(a[i])
        else:
            break
    return pref

words = input().split(', ')
pref = words[0]
for i in range(1, len(words)):
    pref = prefix(pref, words[i])
print("".join(pref))
```
