## Алгоритмы теории чисел

### Решето Эратосфена
---
Алгоритм работает по принципу исключения: мы последовательно вычеркиваем числа, кратные уже найденным простым числам.
Представьте таблицу чисел от \(2\) до \(n\). Мы берем первое невычеркнутое число (это 2) и вычеркиваем все числа, которые делятся на него (4, 6, 8...). Затем переходим к следующему невычеркнутому числу (это 3) и вычеркиваем все кратные ему (6, 9, 12...). Повторяем этот процесс до тех пор, пока не дойдем до конца.

1. Инициализация: Создать массив булевых значений размером от \(0\) до \(n\). Изначально все элементы отмечаются как true (предполагаем, что все числа простые). Индексы 0 и 1 сразу помечаются как false.
2. Фильтрация: Начиная с \(p = 2\), если число \(p\) отмечено как true:
	• Вычеркнуть (отметить как false) все числа в диапазоне от \(p^{2}\) до \(n\) с шагом \(p\) (то есть \(p^2, p^2 + p, p^2 + 2p...\)).
3. Итерация: Перейти к следующему невычеркнутому числу \(p\) и повторить шаг 2.
4. Остановка: Алгоритм можно остановить, как только \(p^2 > n\). Все оставшиеся неотмеченными числа в массиве являются простыми.

> 💡 Оптимизация: Почему мы начинаем вычеркивать с \(p^{2}\), а не с \(2p\)? Потому что все составные числа, меньшие \(p^{2}\), уже будут вычеркнуты предыдущими, более мелкими простыми множителями. По этой же причине цикл внешнего поиска идет только до \(\sqrt{n}\).

- Временная сложность: \(\mathcal{O}(n \log (\log n))\). Это практически линейное время работы, что делает алгоритм невероятно быстрым.
- Сложность по памяти: \(\mathcal{O}(n)\). Требуется хранить массив состояний для каждого числа. При очень больших \(n\) память становится главным ограничением (для решения этой проблемы используют блочное/сегментированное Решето).

```python
N = int(input())
is_prime = [True] * (N+1)
for d in range(2, N+1):
    if is_prime[d]:
        for i in range(d*d, N+1, d):
            is_prime[i] = False
for i in range(2, N+1):
    if is_prime[i]:
        print(i, end=" ")
```
