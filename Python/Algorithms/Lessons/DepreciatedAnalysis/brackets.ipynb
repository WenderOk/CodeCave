### Правильная скобочная последовательность
---

```python
dic = {"(":")", "[":"]", "{":"}"}
stack = []
err = False

s = int(input())

for bracket in s:
    if bracket in dic:
        stack.append(dic[bracket])
    elif bracket in dic.values():
        if not stack or bracket != stack.pop():
            err = True
            break
if stack: err = True
```
