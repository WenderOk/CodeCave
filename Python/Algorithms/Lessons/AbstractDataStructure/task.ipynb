```python
from functools import lru_cache

@lru_cache(maxsize=None)
def F(n):
    if n >= 10000:
        return 1
    if n % 2 == 0:
        return F(n+3)+7
    return F(n+1)-3

for n in range(10000, 10, -10):
    F(n)
print(F(50) - F(57))
# print(F(50))
```

```python
A = [[0,1,0,1,1],
    [1,1,1,1,2],
    [0,1,0,2,2],
    [3,3,1,2,2],
    [0,1,1,0,0]]

def F(x, y):
    if A[x][y] == 2:
        return
    val = A[x][y]
    A[x][y] = 2
    if x + 1 < len(A[0]) and A[y][x+1] == val:
        F(x+1, y)
    if y - 1 < len(A) and A[y-1][x] == val:
        F(x, y+1)
    if x - 1 >= 0 and A[y][x-1] == val:
        F(x-1, y)
    if y + 1 < len(A) and A[y+1][x] == val:
        F(x, y+1)

F(1,3)
print(*A, sep="\n")
```

```python
A = [[0,1,0,1,1],
    [1,1,1,1,2],
    [0,1,0,2,2],
    [3,3,1,2,2],
    [0,1,1,0,0]]

x0, y0 = map(int, input().split())
Q = [[x0,y0]]
color = A[y0][x0]
while len(Q) > 0:
    x, y = Q.pop(0)
    if A[y][x] == color:
        A[y][x] = 2
        if x - 1 >= 0:
            Q.append((x-1, y))
        if y - 1 >= 0:
            Q.append((x, y-1))
        if x + 1 < len(A[0]):
            Q.append((x+1, y))
        if y - 1 < len(A[0]):
            Q.append((x, y+1))


print(*A, sep="\n")
```
